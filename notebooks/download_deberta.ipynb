Test Downloaded of DeBERTa model

```python
import torch
import torch.nn.functional as F
from transformers import AutoTokenizer, AutoModel

# Local path
model_path = DEBERTA_MODEL_PATH

# Load model
print("Loading DeBERTa-base...")
tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModel.from_pretrained(model_path)
model.eval()

# Example SeeAct-style inputs
action_plan = "Click the 'Find Your Truck'"
element_text_1 = "<button>Find Your Truck</button>"
element_text_2 = "<button>Cancel</button>"

def get_embedding(text):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True)
    with torch.no_grad():
        outputs = model(**inputs)
    # Mean pooling
    embeddings = outputs.last_hidden_state.mean(dim=1)
    return embeddings

# Create embeddings
plan_emb = get_embedding(action_plan)
elem1_emb = get_embedding(element_text_1)
elem2_emb = get_embedding(element_text_2)

# Cosine similarity
score1 = F.cosine_similarity(plan_emb, elem1_emb)
score2 = F.cosine_similarity(plan_emb, elem2_emb)

print("\nSimilarity Scores:")
print("Find Your Truck button:", score1.item())
print("Cancel button:", score2.item())

if score1 > score2:
    print("\nCorrect element ranked higher ✅")
else:
    print("\nRanking issue ❌")
```
